```python
from langchain_groq import ChatGroq

llmModel = ChatGroq(model_name="openai/gpt-oss-120b")
```

```python
# llmModel
```

```python
response  = llmModel.invoke("Tell me one fun fact about Prophet Mohamad.")
```

```python
print(response)
```

```python
for chunk in llmModel.stream(
    "Tell me one fun fact about the Mohamad Salah. I need detailed response"
):
    print(chunk.content, end="", flush= True)
```

#### Temperature: more or less creativity

```python
creativeLlmModel = ChatGroq(model_name="openai/gpt-oss-120b", temperature = 0.9)
```

```python
response  = creativeLlmModel.invoke("Tell me one fun fact about Mohamad Salah")
```

```python
print(response)
```

# Chat Model

```python
from langchain_groq import ChatGroq

chatModel = ChatGroq(model="qwen/qwen3.8-27b")
```

```python
messages = [
    ("system", "You are an historian expert in the Kennedy family."),
    ("human", "Tell me one curious thing about JFK."),
]
response = chatModel.invoke(messages)
```

```python
print(response)
```

```python
response
```

```python
response.content
```

```python
response.response_metadata
```

```python
response.description()
```

```python
from langchain_groq import ChatGroq

llamaChatModel = ChatGroq(
    model="llama3-70b-8192"
)
```
