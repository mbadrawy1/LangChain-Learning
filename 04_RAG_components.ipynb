# Embeddings
* Transform the small parts of text in numbers (vectors) that are easily stored and searched by vector databases.

```python
from langchain_huggingface import HuggingFaceEmbeddings

# Standard, fast, and lightweight embedding model
embeddings_model = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
chunks_of_text =     [
        "Hi there!",
        "Hello!",
        "What's your name?",
        "Bond, James Bond",
        "Hello Bond!"
    ]
```

```python
embeddings = embeddings_model.embed_documents(chunks_of_text)
```

# Vector Stores (aka. Vector Databases)
* Store embeddings in a very fast searchable database.

```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter
from langchain_chroma import Chroma

from pathlib import Path
from langchain_core.documents import Document

file_path = "./data/state_of_the_union.txt"

# Read file natively using Python's standard library
text = Path(file_path).read_text(encoding="utf-8")

# Wrap directly into LangChain's Document schema
loaded_document = [Document(page_content=text, metadata={"source": file_path})]

text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap=0)

chunks_of_text = text_splitter.split_documents(loaded_document)

```

```python
len(chunks_of_text)
```

```python
vector_db = Chroma.from_documents(chunks_of_text, embeddings_model)
```

```python
vector_db
```

```python
question = "What did the president say about the John Lewis Voting Rights Act?"

response = vector_db.similarity_search(question)

print(response[0].page_content)
```

# Vector Store as Retriever
* Find the embedding that best answers your question.

```python
from pathlib import Path
from langchain_core.documents import Document

file_path = "./data/state_of_the_union.txt"

# Read file natively using Python's standard library
text = Path(file_path).read_text(encoding="utf-8")

# Wrap directly into LangChain's Document schema
loaded_document= [Document(page_content=text, metadata={"source": file_path})]
```

```python
from langchain_community.vectorstores import FAISS
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap=0)

chunks_of_text = text_splitter.split_documents(loaded_document)

embeddings = embeddings_model
vector_db = FAISS.from_documents(chunks_of_text, embeddings)
```

```python
vector_db
```

```python
retriever = vector_db.as_retriever(search_kwargs={"k": 1})
```

```python
response = retriever.invoke("what did he say about ketanji brown jackson?")
response
```
